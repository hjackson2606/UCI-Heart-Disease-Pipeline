## Pipelines

```python
from sklearn.model_selection import train_test_split

X = heart.drop("num", axis=1)
y = heart["num"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
num_cols = ["age", "trestbps", "chol", "thalch", "oldpeak", "ca"]
```

```python
cat_cols = ["sex", "dataset", "cp", "fbs", "restecg", "exang", "slope", "thal"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

num_pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="median")),
    ("scale", StandardScaler())
])
```

```python
from sklearn.preprocessing import OneHotEncoder

cat_pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("one-hot", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
from sklearn.compose import ColumnTransformer

col_transformer = ColumnTransformer(transformers=[
    ("num_pipeline", num_pipeline, num_cols),
    ("cat_pipeline", cat_pipeline, cat_cols),
], 
    remainder = "drop",
    n_jobs = -1
)
```

## Model Training

### Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold

lr = LogisticRegression(random_state=42)
```

```python
param_grid = {
    "C" : [1,2,3,4]
}
```

```python
CV_lr = GridSearchCV(estimator=lr, param_grid=param_grid, scoring="recall", n_jobs=-1, cv=StratifiedKFold(n_splits=5))
```

```python
pipefinallr = make_pipeline(col_transformer, CV_lr)
pipefinallr.fit(X_train, y_train)
```

```python
from sklearn.metrics import confusion_matrix 
from sklearn.metrics import classification_report

y_pred = pipefinallr.predict(X_test)
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
print(CV_lr.best_params_)
```

### Decision Tree 

```python
from sklearn.tree import DecisionTreeClassifier

dtc = DecisionTreeClassifier()
```

```python
param_grid = {
    "criterion" : ["entropy"],
    "splitter" : ["random"],
    "max_depth" : [8],
    "min_samples_split" : [4],
    "min_samples_leaf" : [6, 8],
    "max_features" : [12, 13]
}
```

```python
CV_dtc = GridSearchCV(estimator=dtc, param_grid=param_grid, scoring="recall", n_jobs=-1, cv=StratifiedKFold(n_splits=5))
```

```python
pipefinaldtc = make_pipeline(col_transformer, CV_dtc)
pipefinaldtc.fit(X_train, y_train)

print("Best CV recall:", CV_dtc.best_score_)
print("Best params:", CV_dtc.best_params_)
model = CV_dtc.best_estimator_
```

```python
# Current Method
bestdtc_model = CV_dtc.best_estimator_
bestdtc_model.predict(X_test)

y_pred = pipefinaldtc.predict(X_test)
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
```
